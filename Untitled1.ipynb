```python
import requests

params = {
    "categories_tags_en": "breakfast-cereals",
    "page_size": 100,
    "fields": "code,product_name,brands,quantity,categories_tags_en,countries_tags,ingredients_text,nutrition_grades,nutriments"
}
```

```python
headers = {
    "User-Agent": "___"
}
```

```python
try:
    response = requests.get("https://world.openfoodfacts.org/api/v2/search", params=params, headers=headers)
    response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)
    payload = response.json()
    products = payload.get('products', [])
    print(f"Pulled {len(products)} products")
except requests.exceptions.RequestException as e:
    print(f"Request failed: {e}")
```

```python
barcodes = [p.get("code") for p in products if p.get("code")]

from pathlib import Path
Path("data/raw").mkdir(parents=True, exist_ok=True)
Path("data/raw/extracted_ids.txt").write_text("\n".join(barcodes), encoding="utf-8")
print(f"Extracted {len(barcodes)} barcodes.")
```

```python
import pandas as pd

df = pd.DataFrame(products)
display(df.head())
```

```python
# Display the DataFrame information to check data types and non-null counts
df.info()
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0])
```

```python
# Drop rows where 'product_quantity' or 'product_quantity_unit' are missing
df.dropna(subset=['product_quantity', 'product_quantity_unit'], inplace=True)
print(f"Dropped rows with missing product quantity/unit. DataFrame now has {len(df)} rows.")

# Re-check missing values after dropping rows
missing_values_after_drop_rows = df.isnull().sum()
print("\nMissing values per column after dropping rows:")
print(missing_values_after_drop_rows[missing_values_after_drop_rows > 0])
```

```python
# Expand the 'nutriments' dictionary column into new columns
# Fill NaN with empty dictionaries for consistent processing
df_nutriments = df['nutriments'].apply(lambda x: x if isinstance(x, dict) else {})

nutriment_keys = sorted(list(set(key for d in df_nutriments for key in d.keys())))

for key in nutriment_keys:
    df[f'nutriment_{key}'] = df_nutriments.apply(lambda x: x.get(key))

# Drop the original 'nutriments' column
df = df.drop(columns=['nutriments'])

print("Expanded 'nutriments' column into individual nutriment columns.")

# Display head with new columns and re-check info
display(df.head())
df.info()
```

```python
# Re-check missing values across all columns after nutriment expansion
missing_values_overall = df.isnull().sum()

print("Missing values per column (showing only columns with missing values):")
print(missing_values_overall[missing_values_overall > 0].sort_values(ascending=False))
```

```python
missing_percentages = df.isnull().sum() / len(df) * 100
columns_to_drop_high_null = missing_percentages[missing_percentages > 90].index.tolist()

print(f"Identified {len(columns_to_drop_high_null)} columns with more than 90% missing values.")
print("Columns to be dropped:", columns_to_drop_high_null)
```

```python
df = df.drop(columns=columns_to_drop_high_null)
print("Dropped columns with high missing value percentages.")

# Re-check missing values after dropping columns
missing_values_after_drop_high_null = df.isnull().sum()
print("\nMissing values per column after dropping high-null columns (showing only columns with missing values):")
print(missing_values_after_drop_high_null[missing_values_after_drop_high_null > 0].sort_values(ascending=False))
```

```python
# Impute 0 for all 'nutriment_' columns that still have missing values
nutriment_cols = [col for col in df.columns if col.startswith('nutriment_')]
for col in nutriment_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(0)

print("Imputed 0 for missing values in 'nutriment_' columns.")

# Re-check missing values after imputation
missing_values_after_imputation = df.isnull().sum()
print("\nMissing values per column after nutriment imputation (showing only columns with missing values):")
print(missing_values_after_imputation[missing_values_after_imputation > 0].sort_values(ascending=False))
```

```python
# Drop 'nutrition_data_per' and 'nutriments_estimated' columns
df = df.drop(columns=['nutrition_data_per', 'nutriments_estimated'])

print("Dropped 'nutrition_data_per' and 'nutriments_estimated' columns.")

# Final check for missing values
missing_values_final = df.isnull().sum()
print("\nMissing values per column after all cleaning steps (showing only columns with missing values):")
print(missing_values_final[missing_values_final > 0])
```

Great! With all missing values handled, our dataset is now clean and ready for further exploration and analysis. The next steps will involve feature engineering, exploring the distribution of key nutritional components, and potentially segmenting products based on their nutritional profiles or other attributes.

```python
# Feature Engineering: Create sugar-to-fat ratio and protein-per-carb ratio
# Replace inf/-inf with NaN before calculating ratios, then fill NaN with 0

df['sugar_to_fat_ratio'] = df['nutriment_sugars_100g'] / df['nutriment_fat_100g']
df['sugar_to_fat_ratio'] = df['sugar_to_fat_ratio'].replace([float('inf'), -float('inf')], pd.NA).fillna(0)

df['protein_per_carb'] = df['nutriment_proteins_100g'] / df['nutriment_carbohydrates_100g']
df['protein_per_carb'] = df['protein_per_carb'].replace([float('inf'), -float('inf')], pd.NA).fillna(0)

print("Created 'sugar_to_fat_ratio' and 'protein_per_carb' features.")
display(df[['product_name', 'nutriment_sugars_100g', 'nutriment_fat_100g', 'sugar_to_fat_ratio',
            'nutriment_proteins_100g', 'nutriment_carbohydrates_100g', 'protein_per_carb']].head())
```

```python
from sklearn.preprocessing import StandardScaler

# Identify numerical columns for clustering (nutriments and engineered features)
numerical_cols = [col for col in df.columns if col.startswith('nutriment_') or col in ['sugar_to_fat_ratio', 'protein_per_carb']]

# Filter out columns that might still be all zeros or have very low variance if any, though imputation should have handled NaNs
# This also implicitly handles any non-numeric columns that might have slipped in
numerical_df = df[numerical_cols].select_dtypes(include=['number'])

# Drop columns that have zero variance (all values are the same, e.g., all 0 after imputation)
zero_variance_cols = numerical_df.columns[numerical_df.var() == 0]
numerical_df = numerical_df.drop(columns=zero_variance_cols)

print(f"Selected {numerical_df.shape[1]} numerical features for clustering.")

# Scale the features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(numerical_df)

# Create a DataFrame for scaled features
scaled_df = pd.DataFrame(scaled_features, columns=numerical_df.columns, index=numerical_df.index)

print("Features scaled successfully. Displaying first 5 rows of scaled data:")
display(scaled_df.head())
```

```python
from sklearn.cluster import KMeans

# Determine optimal number of clusters using Elbow Method (optional, but good practice)
# For simplicity, we'll proceed with an initial assumption of 3 clusters as planned.

# Apply K-Means clustering with 3 clusters
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # n_init to suppress warning
df['cluster'] = kmeans.fit_predict(scaled_df)

print(f"Assigned {k} clusters to the DataFrame.")
print("Cluster distribution:")
display(df['cluster'].value_counts())

# Analyze cluster characteristics by looking at mean values of original (unscaled) features
cluster_centers_original_scale = df.groupby('cluster')[numerical_df.columns].mean()
print("\nMean nutritional values per cluster (original scale):")
display(cluster_centers_original_scale)
```

```python
from sklearn.decomposition import PCA

# Apply PCA to reduce dimensions for visualization
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_df)

# Create a DataFrame with the principal components and cluster labels
pca_df = pd.DataFrame(data=principal_components, columns=['principal_component_1', 'principal_component_2'])
pca_df['cluster'] = df['cluster'].reset_index(drop=True)

plt.figure(figsize=(10, 8))
sns.scatterplot(x='principal_component_1', y='principal_component_2', hue='cluster', data=pca_df,
                palette='viridis', s=100, alpha=0.7, legend='full')
plt.title('Clusters of Breakfast Cereals (PCA)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()

print("Visualized clusters using PCA.")
```

```python
%%writefile src/data_ingestion.py
import requests
import pandas as pd
from pathlib import Path

def fetch_data(category="breakfast-cereals", page_size=100):
    params = {
        "categories_tags_en": category,
        "page_size": page_size,
        "fields": "code,product_name,brands,quantity,categories_tags_en,countries_tags,ingredients_text,nutrition_grades,nutriments"
    }
    headers = {
        "User-Agent": "___" # <<< IMPORTANT: Replace with your actual User-Agent, e.g., "MyFoodAnalysisApp/1.0 (myemail@example.com)"
    }

    try:
        response = requests.get("https://world.openfoodfacts.org/api/v2/search", params=params, headers=headers)
        response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)
        payload = response.json()
        products = payload.get('products', [])
        print(f"Pulled {len(products)} products for category '{category}'.")
        return products
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        return []

def save_raw_data(products, output_path="data/raw/raw_products.json"):
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    pd.DataFrame(products).to_json(output_path, orient="records", indent=4)
    print(f"Raw data saved to {output_path}")

if __name__ == "__main__":
    # Example usage
    products_data = fetch_data()
    if products_data:
        save_raw_data(products_data)
```

```python
%%writefile src/data_cleaning.py
import pandas as pd
from pathlib import Path

def clean_data(input_path="data/raw/raw_products.json", output_path="data/processed/cleaned_products.csv"):
    df = pd.read_json(input_path)
    original_rows = len(df)
    print(f"Loaded {original_rows} rows for cleaning.")

    # Drop rows where 'product_quantity' or 'product_quantity_unit' are missing
    df.dropna(subset=['product_quantity', 'product_quantity_unit'], inplace=True)
    print(f"Dropped rows with missing product quantity/unit. DataFrame now has {len(df)} rows.")

    # Expand the 'nutriments' dictionary column into new columns
    df_nutriments = df['nutriments'].apply(lambda x: x if isinstance(x, dict) else {})

    # Create new DataFrame from expanded nutriments
    expanded_nutriments_df = pd.json_normalize(df_nutriments)
    expanded_nutriments_df.columns = [f'nutriment_{col}' for col in expanded_nutriments_df.columns]

    # Concatenate with original DataFrame and drop original 'nutriments' column
    df = pd.concat([df.reset_index(drop=True), expanded_nutriments_df.reset_index(drop=True)], axis=1)
    df = df.drop(columns=['nutriments'])
    print("Expanded 'nutriments' column into individual nutriment columns.")

    # Drop columns with more than 90% missing values
    missing_percentages = df.isnull().sum() / len(df) * 100
    columns_to_drop_high_null = missing_percentages[missing_percentages > 90].index.tolist()
    if columns_to_drop_high_null:
        df = df.drop(columns=columns_to_drop_high_null)
        print(f"Dropped {len(columns_to_drop_high_null)} columns with >90% missing values.")

    # Impute 0 for all 'nutriment_' columns that still have missing values
    nutriment_cols = [col for col in df.columns if col.startswith('nutriment_')]
    for col in nutriment_cols:
        if pd.api.types.is_numeric_dtype(df[col]): # Only impute numeric columns
            df[col] = df[col].fillna(0)
    print("Imputed 0 for missing values in 'nutriment_' columns.")

    # Drop 'nutrition_data_per' and 'nutriments_estimated' columns as they had remaining NaNs
    cols_to_drop_final = ['nutrition_data_per', 'nutriments_estimated']
    existing_cols_to_drop = [col for col in cols_to_drop_final if col in df.columns]
    if existing_cols_to_drop:
        df = df.drop(columns=existing_cols_to_drop)
        print(f"Dropped final columns: {existing_cols_to_drop}.")

    # Save cleaned data
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(output_path, index=False)
    print(f"Cleaned data saved to {output_path}. Final shape: {df.shape}")
    return df

if __name__ == "__main__":
    # This part will only run when the script is executed directly
    # You might want to run data_ingestion.py first to create the input file
    # For demonstration, ensure a dummy file exists or run ingestion first.
    # For this example, let's assume raw_products.json exists from a previous run or manual creation.
    if not Path("data/raw/raw_products.json").exists():
        print("Error: data/raw/raw_products.json not found. Please run data_ingestion.py first or ensure the file exists.")
    else:
        clean_data()
```

```python
%%writefile src/feature_engineering.py
import pandas as pd
from pathlib import Path

def engineer_features(input_path="data/processed/cleaned_products.csv", output_path="data/processed/features.csv"):
    df = pd.read_csv(input_path)
    print(f"Loaded {len(df)} rows for feature engineering.")

    # Feature Engineering: Create sugar-to-fat ratio and protein-per-carb ratio
    df['sugar_to_fat_ratio'] = df['nutriment_sugars_100g'] / df['nutriment_fat_100g']
    df['sugar_to_fat_ratio'] = df['sugar_to_fat_ratio'].replace([float('inf'), -float('inf')], pd.NA).fillna(0)

    df['protein_per_carb'] = df['nutriment_proteins_100g'] / df['nutriment_carbohydrates_100g']
    df['protein_per_carb'] = df['protein_per_carb'].replace([float('inf'), -float('inf')], pd.NA).fillna(0)

    print("Created 'sugar_to_fat_ratio' and 'protein_per_carb' features.")

    # Save feature-engineered data
    Path(output_path).parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(output_path, index=False)
    print(f"Feature-engineered data saved to {output_path}. Final shape: {df.shape}")
    return df

if __name__ == "__main__":
    if not Path("data/processed/cleaned_products.csv").exists():
        print("Error: data/processed/cleaned_products.csv not found. Please run data_cleaning.py first.")
    else:
        engineer_features()
```

```python
%%writefile src/clustering_model.py
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

def run_clustering(input_path="data/processed/features.csv", output_figure_path="results/clusters_pca.png"):
    df = pd.read_csv(input_path)
    print(f"Loaded {len(df)} rows for clustering.")

    # Identify numerical columns for clustering (nutriments and engineered features)
    numerical_cols = [col for col in df.columns if col.startswith('nutriment_') or col in ['sugar_to_fat_ratio', 'protein_per_carb']]

    numerical_df = df[numerical_cols].select_dtypes(include=['number'])

    # Drop columns that have zero variance
    zero_variance_cols = numerical_df.columns[numerical_df.var() == 0]
    numerical_df = numerical_df.drop(columns=zero_variance_cols)

    print(f"Selected {numerical_df.shape[1]} numerical features for clustering.")

    # Scale the features
    scaler = StandardScaler()
    scaled_features = scaler.fit_transform(numerical_df)
    scaled_df = pd.DataFrame(scaled_features, columns=numerical_df.columns, index=numerical_df.index)
    print("Features scaled successfully.")

    # Apply K-Means clustering with 3 clusters
    k = 3
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # n_init to suppress warning
    df['cluster'] = kmeans.fit_predict(scaled_df)
    print(f"Assigned {k} clusters to the DataFrame.")
    print("Cluster distribution:")
    print(df['cluster'].value_counts())

    # Apply PCA to reduce dimensions for visualization
    pca = PCA(n_components=2)
    principal_components = pca.fit_transform(scaled_df)

    # Create a DataFrame with the principal components and cluster labels
    pca_df = pd.DataFrame(data=principal_components, columns=['principal_component_1', 'principal_component_2'])
    pca_df['cluster'] = df['cluster'].reset_index(drop=True)

    plt.figure(figsize=(10, 8))
    sns.scatterplot(x='principal_component_1', y='principal_component_2', hue='cluster', data=pca_df,
                    palette='viridis', s=100, alpha=0.7, legend='full')
    plt.title('Clusters of Breakfast Cereals (PCA)')
    plt.xlabel('Principal Component 1')
    plt.ylabel('Principal Component 2')
    plt.legend(title='Cluster')
    plt.grid(True)

    Path(output_figure_path).parent.mkdir(parents=True, exist_ok=True)
    plt.savefig(output_figure_path)
    print(f"Visualized clusters using PCA. Plot saved to {output_figure_path}.")

    # Optionally, return the dataframe with clusters or the model
    return df

if __name__ == "__main__":
    if not Path("data/processed/features.csv").exists():
        print("Error: data/processed/features.csv not found. Please run feature_engineering.py first.")
    else:
        run_clustering()
```

```python
%%writefile main.py
from src.data_ingestion import fetch_data, save_raw_data
from src.data_cleaning import clean_data
from src.feature_engineering import engineer_features
from src.clustering_model import run_clustering
from pathlib import Path

def main():
    print("Starting data ingestion...")
    products = fetch_data()
    if not products:
        print("Data ingestion failed or returned no products. Exiting.")
        return
    save_raw_data(products)

    print("\nStarting data cleaning...")
    cleaned_df = clean_data()

    print("\nStarting feature engineering...")
    feature_engineered_df = engineer_features()

    print("\nStarting clustering model...")
    clustered_df = run_clustering()

    print("\nProject pipeline completed successfully!")

if __name__ == "__main__":
    main()
```
